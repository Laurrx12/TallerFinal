```python
class RegistroPartida:

    _instancia = None

    def __init__(self):
        self.historial = []

    @classmethod
    def get_instancia(cls):

        if cls._instancia is None:
            cls._instancia = RegistroPartida()

        return cls._instancia

    def registrar(self, accion):
        self.historial.append(accion)

    def mostrar_historial(self):

        for accion in self.historial:
            print(accion)
```

```python
class Personaje:

    def __init__(self, nombre):
        self.nombre = nombre
        self.hp = self._salud_base()

    def _salud_base(self):
        return 100

    def atacar(self):
        pass
```

```python
class Guerrero(Personaje):

    def _salud_base(self):
        return 150

    def atacar(self):
        return f"{self.nombre} golpea con espada ⚔️"


class Mago(Personaje):

    def _salud_base(self):
        return 80

    def atacar(self):
        return f"{self.nombre} lanza bola de fuego 🔥"


class Arquero(Personaje):

    def _salud_base(self):
        return 100

    def atacar(self):
        return f"{self.nombre} dispara flecha 🏹"
```

```python
class Paladin(Personaje):

    def _salud_base(self):
        return 130

    def atacar(self):
        return f"{self.nombre} bendice con escudo sagrado 🛡️"
```

```python
class FabricaPersonajes:

    _catalogo = {
        "guerrero": Guerrero,
        "mago": Mago,
        "arquero": Arquero,
        "paladin": Paladin
    }

    @classmethod
    def crear(cls, tipo, nombre):

        personaje = cls._catalogo[tipo](nombre)

        RegistroPartida.get_instancia().registrar(
            f"Creado {tipo}: {nombre}"
        )

        return personaje
```

```python
equipo = [

    FabricaPersonajes.crear("guerrero", "Thorin"),

    FabricaPersonajes.crear("mago", "Gandalf"),

    FabricaPersonajes.crear("arquero", "Legolas"),

    FabricaPersonajes.crear("paladin", "Uther")
]
```

```python
for personaje in equipo:
    print(personaje.atacar())
```

```python
class GestorMisiones:

    _instancia = None

    def __init__(self):
        self.misiones = []

    @classmethod
    def get_instancia(cls):

        if cls._instancia is None:
            cls._instancia = GestorMisiones()

        return cls._instancia

    def agregar_mision(self, mision):
        self.misiones.append(mision)

    def listar(self):

        for mision in self.misiones:
            print(mision)
```

```python
from abc import ABC, abstractmethod

# Interfaz de estrategia de defensa
class EstrategiaDefensa(ABC):
    @abstractmethod
    def defender(self, defensor: str) -> str:
        pass

# Estrategias concretas
class EscudoMetalico(EstrategiaDefensa):
    def defender(self, defensor):
        return f"🛡️ {defensor} bloquea con escudo metálico — reduce 30 de daño"

class EvasionRapida(EstrategiaDefensa):
    def defender(self, defensor):
        return f"🏃‍♂️ {defensor} esquiva rápidamente — evita todo el daño"

class BarreraArcana(EstrategiaDefensa):
    def defender(self, defensor):
        return f"✨ {defensor} invoca barrera arcana — reduce 50 de daño mágico"

# Personaje con estrategia de defensa
class PersonajeConDefensa(Personaje):
    def __init__(self, nombre, defensa: EstrategiaDefensa):
        super().__init__(nombre)
        self.defensa = defensa

    def set_defensa(self, nueva_defensa: EstrategiaDefensa):
        self.defensa = nueva_defensa
        print(f"[Strategy] {self.nombre} cambia su estrategia de defensa")

    def defender(self):
        return self.defensa.defender(self.nombre)
```

```python
# Crear personaje con defensa inicial
arthas = PersonajeConDefensa("Arthas", EscudoMetalico())
print(arthas.defender())

# Cambiar estrategia en tiempo de ejecución
arthas.set_defensa(EvasionRapida())
print(arthas.defender())

arthas.set_defensa(BarreraArcana())
print(arthas.defender())
```

```python
# ── Observer: eventos de Personaje ───────────────────────────────────
from abc import ABC, abstractmethod

class ObservadorPersonaje(ABC):
    @abstractmethod
    def actualizar(self, personaje, evento: str): ...

class LogCombate(ObservadorPersonaje):
    def actualizar(self, personaje, evento):
        print(f"  [LOG] {personaje.nombre} → {evento}")

class GestorLogros(ObservadorPersonaje):
    def actualizar(self, personaje, evento):
        if "nivel" in evento:
            print(f"  [LOGRO] {personaje.nombre} alcanzó nivel {personaje.nivel}! 🏆")

class InterfazUI(ObservadorPersonaje):
    def actualizar(self, personaje, evento):
        print(f"  [UI] HP={personaje.salud} Niv={personaje.nivel}")

class SistemaSonido(ObservadorPersonaje):   # 👈 nuevo observador
    def actualizar(self, personaje, evento):
        if "daño" in evento:
            print("[SFX] 💥 sonido_golpe.mp3")
        elif "nivel" in evento:
            print("[SFX] 🎵 level_up.mp3")
        else:
            print("[SFX] 🔔 evento_desconocido.mp3")

# ── Sujeto: Personaje observable ──────────────────────────────────────
class Personaje:
    def __init__(self, nombre: str):
        self.nombre       = nombre
        self.nivel        = 1
        self.salud        = 100
        self._observadores = []

    def suscribir(self, obs):   self._observadores.append(obs)
    def desuscribir(self, obs): self._observadores.remove(obs)

    def _notificar(self, evento: str):
        for obs in self._observadores:
            obs.actualizar(self, evento)

    def subir_nivel(self):
        self.nivel += 1
        self.salud  = 100
        self._notificar(f"sube a nivel {self.nivel}")

    def recibir_danio(self, cantidad: int):
        self.salud -= cantidad
        self._notificar(f"recibe {cantidad} de daño (HP={self.salud})")

# ── Prueba ─────────────────────────────────────────────────────────────
thorin = Personaje("Thorin")
thorin.suscribir(LogCombate())
thorin.suscribir(GestorLogros())
thorin.suscribir(InterfazUI())
thorin.suscribir(SistemaSonido())   # 👈 suscribimos el nuevo observador

print("--- Thorin recibe daño ---")
thorin.recibir_danio(30)

print("--- Thorin sube de nivel ---")
thorin.subir_nivel()

print("--- Desuscribir UI y probar ---")
thorin.desuscribir(thorin._observadores[2])  # quitamos InterfazUI
thorin.recibir_danio(15)
```

```python
# ── Decorator: efectos en ataques del RPG Manager ────────────────────
from abc import ABC, abstractmethod

class Ataque(ABC):
    @abstractmethod
    def ejecutar(self) -> str: ...
    @abstractmethod
    def danio(self) -> int: ...

# Componente concreto
class AtaqueBasico(Ataque):
    def ejecutar(self): return "Ataque básico con espada"
    def danio(self):    return 30

# Decorator base: delega todo al objeto envuelto
class DecoratorAtaque(Ataque):
    def __init__(self, ataque: Ataque):
        self._ataque = ataque
    def ejecutar(self): return self._ataque.ejecutar()
    def danio(self):    return self._ataque.danio()

# Decoradores concretos
class EfectoVeneno(DecoratorAtaque):
    def ejecutar(self): return self._ataque.ejecutar() + " + ☠️ Veneno"
    def danio(self):    return self._ataque.danio() + 15

class EfectoFuego(DecoratorAtaque):
    def ejecutar(self): return self._ataque.ejecutar() + " + 🔥 Fuego"
    def danio(self):    return self._ataque.danio() + 20

class EfectoHielo(DecoratorAtaque):
    def ejecutar(self): return self._ataque.ejecutar() + " + ❄️ Hielo"
    def danio(self):    return self._ataque.danio() + 10

class EfectoRayo(DecoratorAtaque):   # 👈 nuevo decorador
    def ejecutar(self): return self._ataque.ejecutar() + " + ⚡ Rayo"
    def danio(self):    return self._ataque.danio() + 25

# ── Pruebas de composición dinámica ──────────────────────────────────
base = AtaqueBasico()
print(f"[Base]   {base.ejecutar()} → daño: {base.danio()}")

con_fuego = EfectoFuego(base)
print(f"[+Fuego] {con_fuego.ejecutar()} → daño: {con_fuego.danio()}")

combo = EfectoHielo(EfectoVeneno(EfectoFuego(base)))
print(f"[Combo]  {combo.ejecutar()} → daño: {combo.danio()}")

con_rayo = EfectoRayo(base)
print(f"[+Rayo]  {con_rayo.ejecutar()} → daño: {con_rayo.danio()}")

combo_full = EfectoRayo(EfectoHielo(EfectoVeneno(EfectoFuego(base))))
print(f"[Full]   {combo_full.ejecutar()} → daño: {combo_full.danio()}")
```

```python
# ── Template Method: turno de combate ────────────────────────────────
from abc import ABC, abstractmethod

class Personaje(ABC):

    def ejecutar_turno(self, objetivo: str):
        """Template Method — NO se sobreescribe."""
        self.preparar()                       # invariante
        self.atacar(objetivo)                 # variante (abstract)
        self.aplicar_efecto_especial()        # hook (opcional)
        self.cerrar_turno()                   # invariante

    # Pasos invariantes
    def preparar(self):
        print(f"  [{self.__class__.__name__}] Preparando turno...")
    def cerrar_turno(self):
        print(f"  [{self.__class__.__name__}] Turno finalizado.\n")

    # Paso variante — cada subclase lo implementa
    @abstractmethod
    def atacar(self, objetivo: str): ...

    # Hook — las subclases pueden sobreescribirlo o no
    def aplicar_efecto_especial(self):
        pass

# ── Subclases concretas ──────────────────────────────────────────────
class Guerrero(Personaje):
    def atacar(self, objetivo):
        print(f"  ⚔️ Guerrero golpea a {objetivo} — daño: 40")

class Mago(Personaje):
    def atacar(self, objetivo):
        print(f"  🔥 Mago lanza hechizo a {objetivo} — daño: 70")
    def aplicar_efecto_especial(self):
        print("  ✨ El Mago regenera 10 de maná.")

class Arquero(Personaje):
    def atacar(self, objetivo):
        print(f"  🏹 Arquero dispara a {objetivo} — daño: 55")
    def aplicar_efecto_especial(self):
        print("  🎯 Arquero gana ventaja: siguiente ataque crítico.")

class Paladin(Personaje):
    def atacar(self, objetivo):
        print(f"  🛡️ Paladín golpea con mazo a {objetivo} — daño: 50")
    def aplicar_efecto_especial(self):
        print("  🙏 Paladín bendice al equipo: +20 HP para todos.")

# ── Ejecución: misma llamada, diferente comportamiento ────────────────
equipo = [Guerrero(), Mago(), Arquero(), Paladin()]
for personaje in equipo:
    personaje.ejecutar_turno("Dragón")
```
